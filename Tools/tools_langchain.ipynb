## Custom Tools

```python
from langchain_community.tools import tool
```

```python
#creation of function
def multiply(a, b):
    """Multiply two numbers"""
    return a*b
```

```python
#adding user i/p hints
def multiply(a: int, b: int) -> int:
    """Multiply two numbers"""
    return a*b
```

```python
#adding tool decorator
@tool
def multiply(a: int, b: int) -> int:
    """Multiply two numbers"""
    return a*b
```

```python
res=multiply.invoke({'a': 3, 'b': 4})
res
```

```python
print(multiply.name)
print(multiply.description)
print(multiply.args)
```

```python
print(multiply.args_schema.model_json_schema())
```

## Custom Tools: StructuredTools

```python
from langchain_community.tools import StructuredTool
from pydantic import BaseModel, Field
```

```python
class multiply_input(BaseModel):
    a:int = Field(required=True, description='first number to add')
    b:int = Field(required=True, description='second number to add')
```

```python
def multiply_func(a:int, b:int) -> int:
    """multiplication of two numbers"""
    return a*b
```

```python
multiply_tool=StructuredTool.from_function(
    func=multiply_func,
    name="Multiply",
    description="multiplication of given two numbers",
    args_schema=multiply_input
)
```

```python
res=multiply_tool.invoke({'a': 3, 'b': 7})

print(res)
print(multiply_tool.name)
print(multiply_tool.description)
print(multiply_tool.args)
```

## Custom Tools: BaseTool Class

```python
from langchain_community.tools import BaseTool
from typing import Type
```

```python
class multiply_input(BaseModel):
    a:int = Field(required=True, description='first number to add')
    b:int = Field(required=True, description='second number to add')
```

```python
class multiply_tool_base(BaseTool):
    name: str='multiply'
    description: str = "Multiply two numbers"
    args_schema: Type[BaseModel]=multiply_input

    def _run(self, a: int, b: int)->int:
        return a*b
```

```python
multiply_tool_baseTool=multiply_tool_base()
```

```python
res=multiply_tool_baseTool.invoke({'a': 4, 'b': 6})

print(res)
print(multiply_tool_baseTool.name)
print(multiply_tool_baseTool.description)
print(multiply_tool_baseTool.args)
```

## ToolKit

```python
from langchain.tools import tool
```

```python
@tool
def add(a: int, b:int)->int:
    """
    addition of two numbers 
    """
    return a+b
# end def

@tool
def multiplyTK(a: int, b:int)->int:
    """
    multiplication of two numbers 
    """
    return a*b
```

```python
class MathToolKit:
    def get_tools(self):
        return [add, multiplyTK]
```

```python
toolkit=MathToolKit()
tools=toolkit.get_tools()
```

```python
for tool in tools:
    print(f"Tool name: {tool.name} -> does {tool.description}")
```
